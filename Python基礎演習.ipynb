# OpenCVによる画像処理

```python
# NumPyライブラリ（モジュール）
import numpy as np
# OpenCVライブラリ（モジュール）
import cv2
# cv2_imshow関数
from google.colab.patches import cv2_imshow
```

```python
# サンプル画像をダウンロードする
!wget -nc https://github.com/ohta-ee-ous/Lecture_ImageProcessing/releases/download/v1.0.0/sample.jpg
```

```python
# 画像を読み込む
img = cv2.imread('sample.jpg')
```

```python
# 画像を表示する
cv2_imshow(img)
```

```python
# 画像の明るさ変換(値は実数で指定)
dst = img + 100.0
cv2_imshow(dst)
```

```python
# 画像をファイルに保存する
cv2.imwrite('変換後画像.jpg', dst)
```

```python
# 画像をぼかす
# ksize にフィルタサイズ(横幅,縦幅)を指定
dst = cv2.blur(img, ksize=(15, 15))
cv2_imshow(dst)
```

```python
# 画像を反転する
# flipCode に -1, 0, 1 を指定
dst = cv2.flip(img, flipCode=1)
cv2_imshow(dst)
```

```python
# 画像のリサイズ
# fx,fy に横方向,縦方向の倍率を指定
dst = cv2.resize(img, None, fx=1.5, fy=0.5)
cv2_imshow(dst)
```

```python
# 画像のグレイスケール変換
dst = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(dst)
```

# NumPyによる数値計算の基礎

```python
import numpy as np
```

```python
# Python標準のリスト
a = [ 1, 2, 3 ]
print( a )
```

```python
# リストをNumPyのベクトルに変換
b = np.array( a, np.int32 )
print( b )
```

```python
vec1 = np.array( [ 1, 0, 3 ], np.float32 )
vec2 = np.array( [ 4, 3, 2 ], np.float32 )
```

```python
vec3 = vec1 + vec2
print( vec3 )
```

```python
# アダマール積
vec4 = vec1 * vec2
print( vec4 )
```

```python
# 内積
vec5 = np.dot( vec1, vec2 )
print( vec5 )
```

```python
# 外積
vec6 = np.cross( vec1, vec2 )
print( vec6 )
```

```python
# 2次元配列
a = [ [ 1, 2 ],
      [ 0, 3 ] ]
print( a )
```

```python
# NumPyの行列に変換
b = np.array( a, np.int32 )
print( b )
```

```python
mat1 = np.array( [ [1, 2], [3, 4] ], np.float32 )
mat2 = np.array( [ [3, 1], [0, 3] ], np.float32 )
print( 'mat1 =' )
print( mat1 )
print( 'mat2 =' )
print( mat2 )
```

```python
mat3 = mat1 + mat2
print( mat3 )
```

```python
# アダマール積
mat4 = mat1 * mat2
print( mat4 )
```

```python
# 行列積
mat4 = mat1 @ mat2
print( mat4 )
```

```python
mat5 = mat1 * 3
print( mat5 )
```

```python
mat6 = mat1 + 50
print( mat6 )
```

```python
mat7 = mat1 ** 2
print( mat7 )
```
